## As a function

```python
def plot_agg_eco_sex(dataframe = df, variable_one = "Economic Activity", variable_two = "Sex"): # for chat gpt
    df_to_plot = dataframe.groupby([variable_one, variable_two]).size().unstack()
    return df_to_plot.plot(kind = 'bar')
```

```python
plot_agg_eco_sex(dataframe = df, variable_one = "Economic Activity", variable_two = "Residence Type")
```

```python
import matplotlib.pyplot as plt

def plot_agg_eco_sex(
    dataframe: pd.DataFrame,
    variable_one: str = "Economic Activity",
    variable_two: str = "Sex"
) -> plt.Axes:
    """
    Plots a bar chart showing the aggregate count of entries in a DataFrame, 
    grouped by two specified variables.

    Parameters:
    - dataframe: DataFrame
        The input data frame containing the data to plot.
    - variable_one: str, optional
        The column name in the data frame representing the first category (default is "Economic Activity").
    - variable_two: str, optional
        The column name in the data frame representing the second category (default is "Sex").

    Returns:
    - plt.Axes
        The matplotlib Axes object containing the plot.
    """
    df_to_plot = dataframe.groupby([variable_one, variable_two]).size().unstack()
    return df_to_plot.plot(kind='bar')
```

```python
plot_agg_eco_sex(df, 'Economic Activity', 'Sex')
```
