```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('2011.csv', header = 1, index_col = 'Person ID')
```

```python
# Count occurrences of 'Economic Activity' for each 'Sex'
# Assuming 'Sex' is represented as 1 for male and 2 for female, and 'Economic Activity' is a categorical code
counts = df.groupby(['Sex', 'Economic Activity']).size().unstack(fill_value=0)

# Plotting the bar chart
counts.plot(kind='bar', stacked=True)
plt.title('Counts of Economic Activity by Sex')
plt.xlabel('Sex (1 = Male, 2 = Female)')
plt.ylabel('Count')
plt.legend(title='Economic Activity')
plt.show()
```

```python
# Count occurrences of 'Economic Activity' for each 'Sex'
counts = df.groupby(['Sex', 'Economic Activity']).size().unstack(fill_value=0)

# Plotting the bar chart with bars for each Economic Activity grouped together
counts.T.plot(kind='bar', color=['skyblue', 'orange'], width=0.8)
plt.title('Counts of Economic Activity by Sex')
plt.xlabel('Economic Activity')
plt.ylabel('Count')
plt.xticks(rotation=0)  # Rotate x labels if necessary
plt.legend(title='Sex', labels=['Male', 'Female'])  # Adjust labels based on your coding
plt.show()
```
